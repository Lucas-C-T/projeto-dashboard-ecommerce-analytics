Abaixo temos a configuração do ambiente SQL:

```python
import sqlite3
import pandas as pd
```

```python
df_transacoes = pd.read_csv("TB_TRANSACOES_PROJETO_ECOMM.csv", delimiter=';')
df_clientes = pd.read_csv("TB_CLIENTES_PROJETO_ECOMM.csv", delimiter=';')
```

```python
conn = sqlite3.connect('projeto.db')
# Carregar o DataFrame no banco de dados SQLite - criando tb_transacoes e tb_clientes
df_transacoes.to_sql('tb_transacoes', conn, index=False, if_exists='replace')
df_clientes.to_sql('tb_clientes', conn, index=False, if_exists='replace')
```

```python
# Função para executar consultas SQL e retornar o resultado como um DataFrame
def run_query(query):
    return pd.read_sql_query(query, conn)
```

```python
#Colsultar t_transacoes

query = """
SELECT *
FROM tb_transacoes
"""

result_df = run_query(query)
print(result_df)
```

```python
#Colsultar tb_clientes

query = """
SELECT *
FROM tb_clientes
"""

result_df = run_query(query)
print(result_df)
```

```python
query = """
SELECT T.*
FROM tb_transacoes AS T
LEFT JOIN tb_clientes AS C
ON T.Id_client = C.Id_client
WHERE C.Id_client IS NULL
"""

result_df = run_query(query)
print(result_df)
```

```python
query = """
SELECT *
FROM tb_transacoes AS T
INNER JOIN tb_clientes AS C
USING (id_client)

"""

result_df = run_query(query)
print(result_df)
```

Justifique a escolha do JOIN:

O INNER JOIN é o mais adequado porque quer-se analisar apenas os clientes que realmente fizeram transações. Portanto ele garante que cada compra seja vinculada ao cliente correspondente, evitando trazer clientes sem compras (que apareceriam com valores nulos em um LEFT   JOIN).
O RIGHT  JOIN só faria sentido se houvesse necessidade de incluir transações sem clientes associados, o que ocasionaria em problema na integridade nos dados para analise.

Exportando o arquivo como CSV:

```python
result_df.to_csv('dados_ecommerce_final.csv', index=False)
```
